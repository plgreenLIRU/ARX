```python
from ARX.Regressors import Linear
import numpy as np
from matplotlib import pyplot as plt

# Initialise model
m = Linear(N_AR=0, basis='se')
```

```python
X = np.linspace(-5, 5, 30).reshape(-1, 1)
X_star = np.linspace(-30, 30, 200).reshape(-1, 1)
y = np.sin(X).ravel() + 0.2 * np.random.randn(*X.shape).ravel()
centres = np.linspace(-20, 20, 20).reshape(-1, 1)
m.train(X, y, centres=centres, width=5)
```

```python
fig, ax = plt.subplots()

m.model.intercept_ = 0
for i in range(20):
    m.model.coef_ = np.random.randn(len(centres))
    y_pred = m.predict(X_star)
    ax.plot(X_star, y_pred, color='black', alpha=0.2)
ax.plot(m.centres, np.zeros_like(m.centres), 'ro', markersize=5)
ax.grid()
```
